```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from src.data.load_data import load_processed_datasets

from src.data.merge_data import (
    merge_orders_customers,
    merge_order_items,
    merge_products
)
```

```python
datasets = load_processed_datasets()
```

```python
orders_df = datasets["orders_clean"].copy()
customers_df = datasets["customers_clean"].copy()
order_items_df = datasets["order_items_clean"].copy()
products_df = datasets["products_clean"].copy()
payments_df = datasets["payments_clean"].copy()
```

```python
master_df = merge_orders_customers(
    orders_df,
    customers_df
)
```

```python
print("Rows:", len(master_df))
print("Unique Orders:", master_df["order_id"].nunique())
```

```python
master_df = merge_order_items(
    master_df,
    order_items_df
)
```

```python
print("Rows:", len(master_df))

print(
    "Unique Orders:",
    master_df["order_id"].nunique()
)

print(
    "Unique Order-Item combinations:",
    master_df[
        ["order_id", "order_item_id"]
    ].drop_duplicates().shape[0]
)
```

```python
master_df = merge_products(
    master_df,
    products_df
)
```

```python
print("Rows:", len(payments_df))

print(
    "Unique Orders:",
    payments_df["order_id"].nunique()
)
```

```python
payment_counts = (
    payments_df
    .groupby("order_id")
    .size()
)

print(payment_counts.value_counts().sort_index())
```

```python
payment_summary = (
    payments_df
    .groupby("order_id")
    .agg(
        total_payment_value=("payment_value", "sum"),
        payment_count=("payment_sequential", "count"),
        max_installments=("payment_installments", "max")
    )
    .reset_index()
)
```

```python
payment_summary.head()
```

```python
print("Payment summary rows:", len(payment_summary))

print(
    "Unique Orders:",
    payment_summary["order_id"].nunique()
)
```

```python
master_df = pd.merge(
    master_df,
    payment_summary,
    on="order_id",
    how="left"
)
```
